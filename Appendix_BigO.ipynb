```python
# (Emergency Dispatch)

LAST_NAME = "Rojas"
STUDENT_ID = "TUPM-26-1244"
FAVORITE_ARTIST = "Maroon"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - 64 for c in letters)

SIZE = 20

original = [1000 + ID_SUM + i * (SEED_DIGIT + 1) for i in range(SIZE)]
rotation = SEED_DIGIT + 1
rotated = original[rotation:] + original[:rotation]

target = original[(ID_SUM + ARTIST_SUM) % SIZE]
missing = original[-1] + 1


def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)


def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons

        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1, comparisons


print("Seed:", SEED_DIGIT, "| ID Sum:", ID_SUM, "| Artist Sum:", ARTIST_SUM)
print("Original:", original)
print("Rotated :", rotated, f"(first {rotation} moved to end)")

for label, t in [("Target", target), ("Missing", missing)]:
    bi, bc = rotated_binary_search(rotated, t)
    li, lc = linear_search(rotated, t)

    print(f"\n{label} incident: {t}")
    print("Rotated Binary ->", f"position={bi}" if bi != -1 else "NOT FOUND", f"| comparisons={bc}")
    print("Linear         ->", f"position={li}" if li != -1 else "NOT FOUND", f"| comparisons={lc}")

print("\nTime: O(log n) rotated binary, O(n) linear | Space: O(1)")
```

```python
# (University Enrollment)

LAST_NAME = "Rojas"
STUDENT_ID = "TUPM-26-1244"
FAVORITE_ARTIST = "Maroon"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - 64 for c in letters)

dept_a = [ID_SUM + i * (SEED_DIGIT + 1) for i in range(8)]
dept_b = [NAME_LENGTH + i * ARTIST_LENGTH for i in range(10)]

def merge(a, b):
    result, i, j, comparisons = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    return result, comparisons

merged, comparisons = merge(dept_a, dept_b)

if merged != sorted(dept_a + dept_b):
    raise ValueError("Merged list is not in correct order")
if len(merged) != len(dept_a) + len(dept_b):
    raise ValueError("Record count mismatch")

print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated:", merged)
print(f"Records before: {len(dept_a)} + {len(dept_b)} | after: {len(merged)}")
print("Comparisons:", comparisons)
print("Ordering verified: True")
print("Time: O(n + m) | Space: O(n + m)")
```

```python
# (E-Commerce Orders)

import random
import time
LAST_NAME = "Rojas"
STUDENT_ID = "TUPM-26-1244"
FAVORITE_ARTIST = "Maroon"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - 64 for c in letters)

def bubble_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                moves += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, moves

def selection_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            moves += 1
    return a, comparisons, moves

def insertion_sort(data):
    a, comparisons, moves = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            moves += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, moves

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

SIZE = 200
orders = [(ID_SUM * 3 + i * 7 + ARTIST_SUM) % 1000 for i in range(SIZE)]
ascending = sorted(orders)
conditions = {
    "Sorted": ascending,
    "Reverse": ascending[::-1],
    "Random": random.Random(SEED_DIGIT).sample(orders, SIZE),
}
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
}

print(f"Orders generated: {SIZE}")
print(f"{'Condition':<10}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>11}{'Time (s)':>14}")

for cond, data in conditions.items():
    for name, fn in algorithms.items():
        result, elapsed = measure(fn, data)
        if result[0] != ascending:
            raise ValueError(f"{name} failed on {cond}")
        print(f"{cond:<10}{name:<16}{result[1]:>12}{result[2]:>11}{elapsed:>14.8f}")

print("All results verified sorted")
```
